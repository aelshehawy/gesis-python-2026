# Why is pandas AMAZING?

**pandas** is Python's library for working with tabular data - think *"Excel/Stata/R data frames, but programmable"*. It gives us:

1. **Super easy data loading** - CSV, Excel, JSON, databases... one line each.
2. **Easy data cleaning** - remember yesterday's manual CSV wrangling? Never again.
3. **Easy manipulation** - filter, sort, transform thousands of rows at once.
4. **Easy statistics & plotting** - averages, counts, and charts in single commands.

The name comes from "**pan**el **da**ta" - it was literally built for the kind of data social scientists use.

By universal convention, it is imported with the alias `pd`:

````python
# Already installed in Colab - we just import it (with its traditional nickname)
import pandas as pd

print("pandas is ready! Version:", pd.__version__)
````

# The building blocks: Series and DataFrame

pandas has two central objects:

- a **Series** = one labeled column of values
- a **DataFrame** = a whole table (several Series sharing the same row labels - the *index*)

Let's build both from the structures you already know - that's the beautiful part, it's all lists and dictionaries underneath!

````python
# A Series from a LIST - note the index (row labels) pandas adds on the left:
life_exp = pd.Series([79.4, 80.7, 75.6, 82.6])
life_exp
````

*(Notice we didn't need `print()` - like on Day 1 with `1 + 2`, Colab displays the last expression of a cell. For DataFrames this gives pretty formatted tables, so we'll mostly do that today.)*

````python
# A DataFrame from a DICTIONARY: keys become column names, lists become columns
data = {"country": ["Germany", "France", "Poland", "Japan"],
        "life_exp": [79.4, 80.7, 75.6, 82.6],
        "continent": ["Europe", "Europe", "Europe", "Asia"]}

df = pd.DataFrame(data)     # 'df' is THE traditional name for a DataFrame
df
````

````python
# we can do that with a list too 
# Same table, built from a list of rows instead of a dict of columns
rows = [
    ["Germany", 79.4, "Europe"],
    ["France",  80.7, "Europe"],
    ["Poland",  75.6, "Europe"],
    ["Japan",   82.6, "Asia"],
]
listpd=pd.DataFrame(rows, columns=["country", "life_exp", "continent"])
listpd
````

# Filtering: the single most useful pandas skill

*"Only the 2007 rows." "Only Africa." "Only rich countries."* - 90% of data work starts like this.

The recipe: put a **condition** inside the brackets:

```python
df[df["column"] == value]
```

Read it from the inside out: the condition creates a True/False Series (Day 1 booleans - at scale!), and the outer brackets keep the **True** rows.

````python
# Watch the inside part on its own first - a boolean for EVERY row:
gapminder["year"] == 2007
````

````python
# Wrap it in df[...] and we keep only the True rows - one row per country, year 2007:
gm2007 = gapminder[gapminder["year"] == 2007]

print(gm2007.shape)     # 142 rows - our 142 countries
gm2007.head()
````

````python
# All comparison operators from Day 1 work. Life expectancy above 80 in 2007?
gm2007[gm2007["lifeExp"] > 80]
````

````python
# COMBINING conditions - two small differences from Day 1:
#   use & instead of and, | instead of or,
#   and put EACH condition in (parentheses)!
gm2007[(gm2007["continent"] == "Europe") & (gm2007["lifeExp"] > 80)]
````

````python
#save in a new df

df2=gm2007[(gm2007["continent"] == "Europe") & (gm2007["lifeExp"] > 80)]
df2
````

````python
# A handy shortcut: .isin() for "is one of these values"
gapminder[gapminder["country"].isin(["Germany", "Egypt"])].head(20)
````

# Creating new columns

Assign to a column that doesn't exist yet - created on the fly. Calculations happen for **all rows at once**, no loop needed (this is called *vectorization* - more on it in the numpy section):

````python
# Total GDP = GDP per capita × population. 1704 multiplications, one line:
gapminder["gdp_total"] = gapminder["gdpPercap"] * gapminder["pop"]

# In billions, for readability:
gapminder["gdp_billions"] = gapminder["gdp_total"] / 1_000_000_000

gapminder.head(3)
````

````python
# Sanity check - the biggest economies in 2007 (re-filter to get the new columns):
gm2007 = gapminder[gapminder["year"] == 2007]
gm2007.sort_values("gdp_billions", ascending=False)[["country", "gdp_billions"]].head()
````

**Lets create a binary variable/column**

````python
# A BINARY (0/1) column from a condition: keep EVERY row, flag the matches.

# Same ingredients as the filter - but assigned to a new column, not used to drop rows.
gapminder["high_life_europe"] = 0          # everyone starts at 0
gapminder.loc[
    (gapminder["continent"] == "Europe") & (gapminder["lifeExp"] > 80),
    "high_life_europe"
] = 1                                      # flip the matching rows to 1

# 1 = Europe AND life expectancy above 80; 0 = everyone else
gapminder[["country", "year", "continent", "lifeExp", "high_life_europe"]].head()
````

another way of doing the same thing

````python
#we can also do it with another way
gapminder["high_life_europe"] = (
    (gapminder["continent"] == "Europe") & (gapminder["lifeExp"] > 80)
).astype(int)

gapminder
````

lets filter to take a look at what has been tagged as 1

````python
#lets see who has high life expectationcy in europe
gapminder[gapminder["high_life_europe"] == 1]
````

we can also take a look at the unique values and how many of them we do have

````python
#lets see the unique values and how many of them we have
print(gapminder["high_life_europe"].unique())     # [0, 1]
print(gapminder["high_life_europe"].nunique())    # 2
gapminder["high_life_europe"].value_counts()      # how many 0s vs 1s
````

# Missing values

Real data has holes: skipped survey questions, unreported statistics, lost records. pandas marks them as **`NaN`** ("Not a Number").

Our clean Gapminder has none - so I prepared a **realistically messy version** for you. *(It is the same Gapminder data, 2007 slice, with gaps added for practice.)* The toolkit:

| Command | Does |
|---------|------|
| `df.isna()` | True/False map of missingness |
| `df.isna().sum()` | missing count per column |
| `df.dropna()` | drop rows containing any NaN |
| `df.fillna(x)` | replace NaNs with x |

````python
DATA_URL # just to remind ourselves what our url was
````

````python
# Load the messy 2007 data:
messy = pd.read_csv(DATA_URL + "gapminder_messy.csv")

messy.head(20)      # spot the NaN values!
````

````python
# How bad is it? Missing values per column:
messy.isna().sum()
````

````python
messy.isna().head(20) # this shows row by row
````

````python
# Which countries are missing life expectancy? Filter with .isna()!
messy[messy["lifeExp"].isna()]
````

**Why we care - pandas quietly IGNORES NaNs in statistics**

````python
print(f"Mean of the {messy['lifeExp'].count()} observed values: {messy['lifeExp'].mean():.1f}")
print(f"But we have {len(messy)} countries!")
````

- If the missing ones aren't random (say, conflict zones), our mean is biased.
- Missingness is a RESEARCH question, not just a technical one!

````python
# Option A: drop incomplete rows
dropped = messy.dropna()
print(f"Before: {len(messy)} rows → after dropna: {len(dropped)} rows")

````

````python
# dropna(subset=["lifeExp"]) drops only rows where THAT column is missing -
# usually the smarter, more surgical choice:
print(f"Dropping only missing lifeExp: {len(messy.dropna(subset=['lifeExp']))} rows")
````

````python
# Option B: fill the holes - e.g. with the column mean
# (a simple 'imputation'; fine for teaching, but a real analysis needs more thought!)
mean_value = messy["lifeExp"].mean() # we get out the mean value of the column
messy["lifeExp_filled"] = messy["lifeExp"].fillna(mean_value)

print(messy["lifeExp"].isna().sum(), "missing before")
print(messy["lifeExp_filled"].isna().sum(), "missing after")
````

# When files fight back: encodings and European CSVs

Not all data comes from the English-speaking world. 

For example --> A colleague in Cologne exports city data from a **German Excel** and emails it to you *(figures compiled for this course - the official numbers live at [Destatis](https://www.destatis.de), Germany's Federal Statistical Office)*. You do the obvious thing:

````python
# This will NOT work - run it and READ the error, bottom line first:
cities = pd.read_csv(DATA_URL + "german_cities.csv")
````

`UnicodeDecodeError: 'utf-8' codec can't decode byte 0xf6 ...`

Here is what is happening. Computers store text as **numbers**, and an **encoding** is the codebook mapping numbers to characters. 

Today's standard is **UTF-8**, and pandas assumes it. 

But older Windows and Excel setups - still very common in European offices - save files as **Latin-1** (also called ISO-8859-1; close cousin: cp1252). 

In that codebook the byte `0xf6` is the letter **ö** - which is simply invalid in UTF-8. Pandas hit a byte its codebook could not translate and gave up.

The fix: tell pandas which codebook to use.

````python
# Attempt 2: give pandas the right encoding (run it - no error this time!)
cities = pd.read_csv(DATA_URL + "german_cities.csv", encoding="latin-1")
cities.head()
````

No crash - the umlauts are perfect. 

**But look closely at what we got:** everything is squashed into ONE column, and the digits after the decimal commas were split off into a fake column of their own. 

Thios is obviously a silent failure. So its always good to use head() to check your data

German Excel uses the **comma as the decimal sign** (`1,09` instead of `1.09`) - so it cannot also use the comma to separate columns. German CSVs therefore separate columns with **semicolons**. pandas, expecting commas, found "columns" in the wrong places. 

Tell it the real separator with `sep`:

````python
# Attempt 3: right encoding AND the right separator
cities = pd.read_csv(DATA_URL + "german_cities.csv", encoding="latin-1", sep=";")
cities.head()
````

````python
# Looks perfect... but check the TYPES. population_millions is 'object' (text!),
# because pandas does not recognize "1,09" as a number:
print(cities.dtypes)
````

````python
# Attempt 4 - the full solution: encoding + separator + decimal sign
cities = pd.read_csv(DATA_URL + "german_cities.csv",
                     encoding="latin-1",   # the codebook for ö, ü, ß ...
                     sep=";",              # German CSVs separate with semicolons...
                     decimal=",")          # ...because the comma is the decimal sign

print(cities.dtypes)                       # population is a proper float now!
cities.sort_values("population_millions", ascending=False).head()
````

# A pinch of numpy

**numpy** ("numerical Python") is the mathematical engine **underneath pandas** - every DataFrame column is secretly a numpy *array*. You will constantly see it imported as `np` in code online, so let's demystify it:

````python
import numpy as np    # the traditional alias, like pd for pandas

# A numpy array from a list - looks like a list, but supercharged for math:
values = np.array([79.4, 80.7, 75.6, 82.6])
print(values)
print(type(values))
````

Below, this is applied to every element at once!

````python
# The superpower: VECTORIZED math - operations hit ALL elements at once.
# Remember writing loops for this yesterday?
print(values + 10)       # add 10 to every element
print(values * 2)        # double every element
print(values > 79)       # compare every element - hello, filtering logic!
````

````python
# numpy's statistics (pandas uses these internally):
print(np.mean(values))
print(np.median(values))
print(np.std(values))       # standard deviation
print(np.round(values))     # round every element
````

````python
# The pandas ↔ numpy bridge in both directions:
life_exp_column = gapminder["lifeExp"]

as_array = life_exp_column.to_numpy()    # Series → numpy array
print(as_array[:5])                      # (slicing works, of course)

# and numpy functions accept pandas columns directly:
print(np.mean(life_exp_column))
````

**When do YOU need numpy directly?** Mostly: random numbers for simulations, mathematical functions (`np.log`, `np.sqrt` - e.g. logging GDP before a regression), and fancier array math. For this course: *know it exists, recognize `np.`, remember it's what makes pandas fast.*
