## Exercise 5 - New columns

Using `gm2007`:

1. Create a column `gdp_total` = `gdpPercap` × `pop`.
2. Create a column `pop_millions` = `pop` / 1,000,000, for readability.
3. Sort by `gdp_total` descending and show the top 5 (columns: `country`, `pop_millions`, `gdp_total`).
4. (Bonus) Create a boolean column `rich` that is `True` where `gdpPercap` is above 12,000. What **share** of 2007 countries is rich? *(Fun fact: `.mean()` on a True/False column gives exactly the share of Trues - think about why!)*

**Solution:**

```python
# 1. + 2. Vectorized: computed for all 142 rows at once
# (pandas may show a 'SettingWithCopyWarning' here because gm2007 is a filtered
#  slice - for our purposes it is safe to ignore; .copy() when filtering avoids it.)
gm2007 = gapminder[gapminder["year"] == 2007].copy()
gm2007["gdp_total"] = gm2007["gdpPercap"] * gm2007["pop"]
gm2007["pop_millions"] = gm2007["pop"] / 1_000_000

# 3. The biggest economies
gm2007.sort_values("gdp_total", ascending=False)[["country", "pop_millions", "gdp_total"]].head()
```

```python
# 4. (Bonus) A boolean column...
gm2007["rich"] = gm2007["gdpPercap"] > 12000

# ...whose mean is the SHARE of True values: True counts as 1, False as 0,
# so the mean = (number of Trues) / (number of rows). A lovely trick!
print(f"Share of rich countries in 2007: {gm2007['rich'].mean():.1%}")
```

## Exercise 7 - Missing values

1. Load `gapminder_messy.csv` into a DataFrame called `messy` (same `DATA_URL` trick).
2. How many missing values does **each column** have?
3. Show the rows where `gdpPercap` is missing. How many are there?
4. Create `complete`: the messy data with all incomplete rows **dropped**. Compare the number of rows before/after.
5. Fill the missing `pop` values with the column's **median** into a new column `pop_filled`. Verify there are no missing values left in it.
6. (Bonus) In a comment: why might dropping incomplete rows **bias** an analysis of global health? (One sentence - think about *which* countries tend to have missing data!)

**Solution:**

```python
# 1. Load the messy file
messy = pd.read_csv(DATA_URL + "gapminder_messy.csv")

# 2. Missing per column
print(messy.isna().sum())
```

```python
# 3. Rows with missing GDP
missing_gdp = messy[messy["gdpPercap"].isna()]
print(len(missing_gdp), "rows have no GDP value:")
missing_gdp
```

```python
# 4. Drop all incomplete rows
complete = messy.dropna()
print(f"{len(messy)} rows before → {len(complete)} after dropna")
```

```python
# 5. Median imputation into a NEW column (original stays untouched - good practice)
median_pop = messy["pop"].median()
messy["pop_filled"] = messy["pop"].fillna(median_pop)

print(messy["pop_filled"].isna().sum(), "missing values left")   # 0
```

6. (Bonus) Data is rarely missing at random: countries in conflict or with weak statistical agencies are *exactly* the ones with missing health data - dropping them systematically removes the worst-off cases and makes global health look **better** than it is. *(This is why missingness is a research question!)*

## Exercise 8 - A pinch of numpy

1. Convert the `lifeExp` column of `gm2007` to a numpy array called `arr`.
2. Print its mean, median, and standard deviation - using `np.` functions.
3. In one vectorized line: how many countries have a value above 75? *(Hint: comparisons give True/False arrays, and `.sum()` counts the Trues!)*
4. Create `log_gdp` - the natural log (`np.log`) of the `gdpPercap` column - and print its first 5 values. *(Economists log GDP constantly; this afternoon you'll see it makes plots much nicer, too.)*

**Solution:**

```python
gm2007 = gapminder[gapminder["year"] == 2007]

# 1. Series → numpy array
arr = gm2007["lifeExp"].to_numpy()

# 2. numpy statistics
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))

# 3. The comparison gives an array of True/False; .sum() counts the Trues
print((arr > 75).sum(), "countries above 75 years")

# 4. Vectorized log of the whole column
log_gdp = np.log(gm2007["gdpPercap"])
print(log_gdp.head())
```

## Exercise 9 (Bonus) - Mini research question

**"Did the global life-expectancy gap between continents narrow between 1952 and 2007?"**

1. Filter to 1952. Compute the mean `lifeExp` **for Africa** and **for Europe** (two filters + `.mean()`).
2. Repeat for 2007.
3. Compute the Africa-Europe gap in both years and print a mini-conclusion sentence with both numbers.
4. Save your 2007 Africa DataFrame as `africa_2007.csv` and check it appears in the Files sidebar.

*(This afternoon you'll learn `.groupby()`, which answers question 1-2 for ALL continents in one line. Feel free to peek ahead...)*

**Solution:**

```python
# 1. 1952: mean by continent, via two filters
gm1952 = gapminder[gapminder["year"] == 1952]
africa_1952 = gm1952[gm1952["continent"] == "Africa"]["lifeExp"].mean()
europe_1952 = gm1952[gm1952["continent"] == "Europe"]["lifeExp"].mean()

# 2. Same for 2007
gm2007 = gapminder[gapminder["year"] == 2007]
africa_2007 = gm2007[gm2007["continent"] == "Africa"]["lifeExp"].mean()
europe_2007 = gm2007[gm2007["continent"] == "Europe"]["lifeExp"].mean()

# 3. The gaps
gap_1952 = europe_1952 - africa_1952
gap_2007 = europe_2007 - africa_2007

print(f"1952: Europe {europe_1952:.1f} vs Africa {africa_1952:.1f} → gap {gap_1952:.1f} years")
print(f"2007: Europe {europe_2007:.1f} vs Africa {africa_2007:.1f} → gap {gap_2007:.1f} years")
print("Both continents improved - but the gap WIDENED slightly. Life got better AND stayed unequal.")
```

```python
# 4. Export & check the sidebar
africa_df = gm2007[gm2007["continent"] == "Africa"]
africa_df.to_csv("africa_2007.csv", index=False)
print("Saved")
```
