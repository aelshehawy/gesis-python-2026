# Variables: giving things a name

A **variable** stores a value under a name, using `=` (the *assignment operator*).

Think of it as a labeled box: you put something in, and later you use the label to get it back.

**Naming rules:**
- Must start with a letter (or `_`), and can contain letters, numbers, and underscores.
- No spaces! Use underscores instead: `my_variable`.
- Case matters: `Age` and `age` are different variables.
- Choose **descriptive names**: `population` is much better than `p` or `x2`.

````python
# Create a variable called 'course' and store some text in it
course = "Introduction to Python"

# Now the name 'course' contains that text - let's print it
print(course)
````

````python
# Variables can hold numbers too
participants = 25

print(participants)
````

````python
# You can change (re-assign) a variable at any time - the old value is overwritten
participants = 25
print(participants)

participants = 27   # two more people joined late
print(participants)
````

````python
# Variables can be combined in calculations
coffee_per_person = 2                          # cups per person per day
total_coffee = participants * coffee_per_person  # 27 * 2

print(total_coffee)  # cups of coffee this course needs per day
````

# Data types

Every value in Python has a **type**. The four essential ones:

| Type | Name | Example | Used for |
|------|------|---------|----------|
| `int` | integer | `42` | whole numbers (counts, years) |
| `float` | floating point | `3.14` | decimal numbers (rates, percentages) |
| `str` | string | `"Berlin"` | text |
| `bool` | boolean | `True`, `False` | yes/no, on/off |

You can always ask Python what type something is with the `type()` function.

````python
# An integer: a whole number
year = 2026
print(type(year))
````

````python
# A float: a number with a decimal point
life_expectancy = 82.6      # Japan, 2007 - highest in the data we will use!
print(type(life_expectancy))
````

````python
# A string: text in quotes
country = "Japan"
print(type(country))
````

````python
# A boolean: True or False (capital first letter, no quotes!)
is_online_course = True
print(type(is_online_course))
````

````python
# Watch out: numbers in quotes are STRINGS, not numbers!
number_real = 2026        # an int - you can calculate with this
number_text = "2026"      # a str - this is text that happens to look like a number

print(type(number_real))
print(type(number_text))
````

# Mathematical operations

Python knows all the classics - and a few extras that are very useful:

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `+` | addition | `7 + 2` | `9` |
| `-` | subtraction | `7 - 2` | `5` |
| `*` | multiplication | `7 * 2` | `14` |
| `/` | division | `7 / 2` | `3.5` |
| `//` | **integer** division (drops the decimals) | `7 // 2` | `3` |
| `%` | modulo (the remainder) | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

````python
# The basics
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)    # normal division - always gives a float!
````

````python
# The interesting ones:
print(7 // 2)   # integer division: how many WHOLE times does 2 fit into 7?
print(7 % 2)    # modulo: what REMAINS after that division?
print(7 ** 2)   # 7 to the power of 2
````

````python
# Python respects the order of operations (BODMAS/PEMDAS) - use parentheses to control it
print(2 + 3 * 4)      # multiplication first: 2 + 12
print((2 + 3) * 4)    # parentheses first: 5 * 4
````

### A real example: population density

Let's combine variables and math. Germany has about 82.4 million inhabitants and an area of about 357,600 km²:

````python
# Store what we know in clearly named variables
population = 82400000        # people (Germany, 2007)
area_km2 = 357600            # square kilometers

# Calculate people per square kilometer
density = population / area_km2

print(density)   # ≈ 230 people per km²
````

# Boolean operations: asking yes/no questions

**Comparison operators** compare two values and answer `True` or `False`:

| Operator | Question |
|----------|----------|
| `==` | is equal to? ( TWO equals signs - one `=` assigns a variable!) |
| `!=` | is NOT equal to? |
| `>` , `<` | greater / smaller than? |
| `>=` , `<=` | greater / smaller than or equal? |

These will become essential on Day 2 (conditions) and Day 3 (filtering data).

````python
# Life expectancy in 2007, from the data we will analyze on Day 3:
life_exp_japan = 82.6
life_exp_germany = 79.4

print(life_exp_japan > life_exp_germany)    # is Japan's higher?
print(life_exp_japan == life_exp_germany)   # are they equal?
print(life_exp_japan != life_exp_germany)   # are they different?
````

````python
# Combine conditions with 'and' / 'or', flip them with 'not':
age = 30

print(age > 18 and age < 65)   # BOTH must be True
print(age < 18 or age > 25)    # at least ONE must be True
print(not age > 18)            # 'not' turns True into False and vice versa
````

# Converting between types

Sometimes you need to transform a value from one type to another:

- `int(x)` → to integer
- `float(x)` → to float
- `str(x)` → to string

Why does this matter? Because mixing types can cause errors - let's see.

````python
# This causes an error - you cannot glue text and a number together with +
age = 30
print("My age is " + age)
````

`TypeError: can only concatenate str (not "int") to str` - Python refuses to add a string and an integer.

**The fix:** convert the number to a string first.

````python
# Convert the int to a str, and the + works:
age = 30
print("My age is " + str(age))
````

````python
# More conversions:
print(int("2026"))     # string → int  (works because "2026" looks like a number)
print(float("3.14"))   # string → float
print(int(82.6))       # float → int - careful, this CUTS OFF the decimals (no rounding!)
print(round(82.6))     # if you want proper rounding, use round()
````

# Understanding Python strings

Time to get comfortable with text - as social scientists, an enormous part of our data is text!

Strings are created with quotes (single or double - just be consistent):

````python
# Two ways to write the same string
city1 = "Cologne"
city2 = 'Cologne'
print(city1 == city2)   # identical!

# Why have both? So you can put one kind of quote INSIDE the other:
sentence = "GESIS is located in Cologne, or 'Köln' in German"
print(sentence)
````

````python
# len() tells you how many characters a string has (including spaces!)
slogan = "Python for social science"
print(len(slogan))
````

````python
# + glues strings together (concatenation), * repeats them
first_name = "Ada"
last_name = "Lovelace"

full_name = first_name + " " + last_name    # don't forget the space!
print(full_name)

print("na" * 8 + " Batman!")             # strings can be multiplied by ints
````

## f-strings: the elegant way to combine text and variables

Constantly writing `str()` and `+` gets tedious. **f-strings** solve this beautifully:

- Put an `f` **before** the opening quote.
- Put variables inside curly braces `{}` - Python fills in their values automatically.

This is the modern standard, and we will use it for the rest of the course.

````python
country = "Vietnam"
life_1952 = 40.4
life_2007 = 74.2

# An f-string: variables go directly inside {curly braces}
print(f"Life expectancy in {country} rose from {life_1952} to {life_2007} years!")
````

````python
# Bonus trick: format numbers inside the braces.
# :.1f means "show as a decimal number with 1 digit after the point"
gdp_per_capita = 31656.06806

print(f"Ireland's GDP per capita in 2007 was {gdp_per_capita:.1f} dollars")
print(f"Or with 2 decimals: {gdp_per_capita:.2f} dollars")

# ,.0f adds thousands separators - great for big numbers like populations
population = 82400996
print(f"Germany had {population:,.0f} inhabitants in 2007")
````

````python
# Two more print() tricks: 'sep' controls what goes BETWEEN items,
# 'end' controls what comes at the END (default: a line break)
print("Day", "1", "Session", "1")               # default: separated by spaces
print("Day", "1", "Session", "1", sep="**")     # separated by **
print("25", "08", "2026", sep="-")              # useful for dates!
````

## String methods: your text toolbox

A **method** is an action attached to an object, called with a dot: `text.upper()`.

The most useful string methods:

| Method | What it does |
|--------|--------------|
| `.upper()` / `.lower()` | ALL CAPS / all lowercase |
| `.title()` | First Letter Of Each Word Capitalized |
| `.strip()` | removes spaces at the start & end |
| `.replace(old, new)` | replaces text |
| `.count(x)` | counts how often `x` appears |
| `.split()` | splits the string into a list of words (→ this afternoon!) |

````python
institute = "gesis leibniz institute for the social sciences"

print(institute.upper())     # SHOUT IT
print(institute.title())     # Name-Style Capitalization
print(institute.count("s"))  # how many letter s?
````

````python
# .strip() - messy data's best friend. Survey responses often have stray spaces:
messy_answer = "   Political Science   "
print(messy_answer)                    # you can't see the spaces, but they are there
print(len(messy_answer))               # 24 characters?!

clean_answer = messy_answer.strip()    # spaces at the edges: gone
print(clean_answer)
print(len(clean_answer))               # 17 - much better
````

````python
# .replace() swaps text - here we anonymize a sentence:
comment = "I think Angela should win the election"
anonymous = comment.replace("Angela", "[NAME]")
print(anonymous)
````

````python
# .split() chops a string into words - a first taste of text analysis!
speech_line = "We choose to go to the Moon in this decade"

words = speech_line.split()    # splits at every space
print(words)
print(f"That line has {len(words)} words")
````

````python
# Methods can be CHAINED - the output of one becomes the input of the next:
messy = "   pOLITICAL sCIENCE   "
print(messy.strip().lower().title())   # strip → lowercase → title-case, in one line
````
