# Lists

A list is an **ordered, changeable** collection, written with **square brackets** `[ ]`, items separated by commas.

A list can hold anything - numbers, strings, booleans... even other lists.

````python
# A list of strings - for example, the countries in our study:
countries = ["Germany", "France", "Poland", "Italy", "Spain"]
print(countries)

# len() works on lists too: how many items?
print(len(countries))
````

````python
# Lists can mix types (though usually you keep them consistent)
from typing import Any


mixed_list = ["Berlin", 3.5, 2026, True]
print(mixed_list)
print(type(mixed_list))
````

## Indexing: grabbing single items

Every item has a position - its **index**.

> **VERY IMPORTANT:** Python starts counting at **0**, not 1!
> The first item is `[0]`, the second is `[1]`, and so on. This trips up everyone at first (especially if you come from R) - after this week it will feel natural.

Negative indices count **from the end**: `[-1]` is the last item.

````python
countries = ["Germany", "France", "Poland", "Italy", "Spain"]

#              index 0     index 1    index 2   index 3   index 4
#       also:  index -5    index -4   index -3  index -2  index -1

print(countries[0])     # the FIRST item
print(countries[1])     # the second item
print(countries[-1])    # the LAST item - no need to know how long the list is!
print(countries[-2])    # second to last
````

````python
# Asking for a position that does not exist gives an error - run and read it!
countries = ["Germany", "France", "Poland", "Italy", "Spain"]
print(countries[5])   # there is no index 5 - only 0,1,2,3,4
````

`IndexError: list index out of range` - Python says: *"this list has no item at position 5."* Remember: 5 items → highest index is **4**.

## Slicing: grabbing several items

`list[start:stop]` gives you a piece of the list, from `start` **up to but NOT including** `stop`.

Leaving out `start` means "from the beginning"; leaving out `stop` means "to the end".

````python
countries = ["Germany", "France", "Poland", "Italy", "Spain"]

print(countries[1:3])    # index 1 and 2 - the 3 is NOT included!
print(countries[:2])     # from the start up to (not incl.) index 2
print(countries[2:])     # from index 2 to the very end
print(countries[-2:])    # the last two items
````

````python
# Strings are sequences too - indexing and slicing work exactly the same way!
word = "GESIS"

print(word[0])      # first character
print(word[-1])     # last character
print(word[0:3])    # first three characters
````

## Changing lists

Lists are **mutable** - you can modify them after creation:

| Action | Code |
|--------|------|
| change an item | `my_list[0] = "new value"` |
| add to the end | `my_list.append(x)` |
| insert at a position | `my_list.insert(position, x)` |
| remove by value | `my_list.remove(x)` |
| remove (and get) the last item | `my_list.pop()` |

````python
# Change an item by assigning to its index
countries = ["Germany", "France", "Poland", "Italy", "Spain"]
countries[1] = "Netherlands"     # France leaves the study, Netherlands joins
print(countries)
````

````python
# .append() adds ONE item to the END - you will use this constantly!
countries = ["Germany", "France", "Poland"]
print(countries)

countries.append("Portugal")     # note: this CHANGES the list directly
print(countries)
````

````python
# Start empty, then fill up - an extremely common pattern (essential tomorrow!)
participants = []                  # an empty list
print(participants)

participants.append("Amina")       # people register one by one...
participants.append("Boris")
participants.append("Carla")
print(participants)
````

````python
# .insert() puts an item at a specific position (everything else shifts right)
waitlist = ["Boris", "Carla"]
waitlist.insert(0, "Amina")        # Amina jumps to the front of the queue
print(waitlist)
````

````python
# .remove() deletes by VALUE...
letters = ["a", "b", "c", "a"]
letters.remove("a")
print(letters)     # careful: only the FIRST "a" was removed!
````

````python
# .pop() removes the LAST item and hands it to you
beach = ["sun", "sand", "waves"]
last_item = beach.pop()

print(last_item)   # what we took out
print(beach)       # what remains

# You can also pop a specific position: beach.pop(0) removes the first item
````

## Sorting and quick statistics

Lists of numbers come with free statistics: `min()`, `max()`, `sum()` - and `sorted()` puts any list in order.

````python
# Ages of participants in a (very small) survey:
ages = [23, 16, 14, 28, 19, 11, 38]

print(min(ages))    # youngest
print(max(ages))    # oldest
print(sum(ages))    # total

# The average: total divided by the number of people
average = sum(ages) / len(ages)
print(f"Average age: {average:.1f} years")
````

````python
ages = [23, 16, 14, 28, 19, 11, 38]

# sorted() returns a NEW, sorted list (the original stays untouched)
print(sorted(ages))                   # ascending
print(sorted(ages, reverse=True))     # descending
print(ages)                           # original: unchanged!

# Works on strings too - alphabetical order:
names = ["Carla", "Amina", "Boris"]
print(sorted(names))
````

````python
# 'in' asks: is this item in the list? (gives a boolean - remember this morning!)
ages = [23, 16, 14, 28, 19, 11, 38]

print(38 in ages)        # True
print(99 in ages)        # False
print(99 not in ages)    # True - 'not in' also works and reads nicely
````

````python
# Lists inside lists! Each inner list is one row of data - a mini-dataset:
#            [country,    capital,   population in millions]
countries = [["Germany",  "Berlin",  82.4],
             ["France",   "Paris",   68.2],
             ["Poland",   "Warsaw",  36.7]]

print(countries[0])        # the whole first inner list (Germany's row)
print(countries[0][1])     # first row → its item at index 1 → "Berlin"
print(countries[2][2])     # Poland's population
````

````python
# we can also amend a list of lists
countries[0][0] = "Deutschland"
print(countries)
````

````python
countries[1].remove(68.2)   # remove by value
print(countries)
````

````python
del countries[1][1] # remove by position         # remove by position
print(countries)
````

````python
countries[2][2] = int(countries[2][2]) # turn into an interger
print(countries)
````

# Tuples

A tuple is like a list, but **immutable** - once created, it can **never be changed**. Written with **round parentheses** `( )`.

**Why would we want that?** For data that *should not* change: coordinates, (day, month, year) dates, fixed category codes. If it's in a tuple, you can be confident nobody accidentally modified it.

````python
# GESIS headquarters in Cologne, as (latitude, longitude):
gesis_location = (50.9375, 6.9603)

# Reading works exactly like lists:
print(gesis_location[0])    # latitude
print(gesis_location[1])    # longitude
````

````python
# But CHANGING is forbidden - run it and read the error:
gesis_location = (50.9375, 6.9603)
gesis_location[0] = 99.9
````

`TypeError: 'tuple' object does not support item assignment` - exactly the point of a tuple: **read-only data**.

*(Tiny quirk: a tuple with ONE item needs a trailing comma: `solo = ("Berlin",)` - without the comma Python just sees parentheses around a string.)*

````python
solo = ("Berlin")
type(solo)
````

````python
solo = ("Berlin",)
type(solo)
````

# Dictionaries

The star of today. A dictionary stores **key-value pairs**, written with **curly braces** `{ }`:

```python
my_dict = {"key1": value1, "key2": value2}
```

Like a real dictionary: you look up a *word* (the key) and get its *definition* (the value). Keys must be **unique**.

First, let's see the problem dictionaries solve. Suppose we note each country's life expectancy in a **list**:

````python
# This works, but it is fragile and hard to use:
life_exp_list = ["Germany", 79.4, "France", 80.7, "Poland", 75.6]

# To find Poland's value you must REMEMBER it comes right after "Poland"... yuck.
print(life_exp_list[5])
````

````python
# The dictionary way: each country (key) points to its value
life_exp = {"Germany": 79.4,
            "France": 80.7,
            "Poland": 75.6}

print(life_exp)

# Now we look things up BY NAME - readable and safe:
print(life_exp["Poland"])
````

````python
# Add a new pair: just assign to a new key
life_exp = {"Germany": 79.4, "France": 80.7, "Poland": 75.6}
life_exp["Spain"] = 80.9
print(life_exp)

# Update an existing key the same way (keys are unique - the old value is replaced)
life_exp["Germany"] = 81.1
print(life_exp)
````

````python
# Asking for a key that does not exist → KeyError:
life_exp = {"Germany": 79.4, "France": 80.7}
print(life_exp["Italy"])
````

````python
# The polite alternative: .get() returns None (or a default you choose) instead of crashing
life_exp = {"Germany": 79.4, "France": 80.7}

print(life_exp.get("Italy"))                  # None - no crash!
print(life_exp.get("Italy", "no data"))    # or your chosen fallback
````

````python
# Three views into a dictionary - these will be very useful with loops tomorrow:
life_exp = {"Germany": 79.4, "France": 80.7, "Poland": 75.6}

print(life_exp.keys())      # just the keys
print(life_exp.values())    # just the values
print(life_exp.items())     # (key, value) pairs

# 'in' checks the KEYS of a dictionary:
print("Germany" in life_exp)   # True
print(79.4 in life_exp)        # False - 79.4 is a value, not a key!
````

````python
# If you want to know if the value exists in the dictionary
print(80.7 in life_exp.values())
print(life_exp["Poland"] == 79.4)   # True or False
````

## Nested dictionaries

Values can be **anything** - including lists or other dictionaries. This is how you represent richer records (and it is exactly how most web data - JSON - is structured!):

````python
# One dictionary per country, inside an outer dictionary:
countries_info = {
    "Germany": {"capital": "Berlin",  "population_m": 82.4, "continent": "Europe"},
    "Egypt":   {"capital": "Cairo",   "population_m": 80.3, "continent": "Africa"},
    "Japan":   {"capital": "Tokyo",   "population_m": 127.5, "continent": "Asia"},
}

# Work from the outside in, one pair of brackets at a time:
print(countries_info["Egypt"])                   # Egypt's whole inner dictionary
print(countries_info["Egypt"]["capital"])        # → "Cairo"
print(countries_info["Japan"]["population_m"])   # → 127.5
````

````python
# in case youd wantg to make a change in the dictionary
countries_info["Germany"]["continent"] = "Euro"
print(countries_info)
````

- Renaming a key works differently from changing a value 
- You can't reassign a key in place
- The trick is to add the new key and remove the old one. 

````python
countries_info["Deutschland"] = countries_info["Germany"]
print(countries_info)
````

````python
del countries_info["Germany"]
print(countries_info)
````

Values can also be lists!

````python
# Values can also be lists - e.g. name variations for text searches
# (super useful later if you ever search text data for country mentions!)
korea_variants = {
    "South Korea": ["Korea", "South Korea", "South-Korea", "Republic of Korea"],
    "North Korea": ["North Korea", "DPRK", "Democratic People's Republic of Korea"],
}

print(korea_variants["South Korea"])        # the whole list
print(korea_variants["South Korea"][0])     # dictionary lookup, THEN list index
````

# Sets

A set is an **unordered** collection of **unique** items - duplicates are automatically removed. Written with curly braces, but *without* colons: `{1, 2, 3}`.

Their superpower: instant answers to *"which unique values do I have?"* and comparisons between groups.

````python
# Duplicates disappear on creation:
responses = {"yes", "no", "yes", "maybe", "yes"}
print(responses)     # each value only once (and in no particular order!)
````

````python
# The classic move: set() de-duplicates a list
votes = ["SPD", "CDU", "SPD", "Greens", "CDU", "SPD", "Greens"]

unique_parties = set(votes)
print(unique_parties)               # which parties appeared at all?
print(len(unique_parties))          # how many DIFFERENT parties?

# and list() turns it back into a list if needed:
unique_list = list(unique_parties)
print(unique_list)
````

````python
# Sets can be modified - add / update / discard:
parties = {"SPD", "CDU"}
parties.add("Greens")               # add one item
print(parties)

parties.update({"FDP", "Linke"})    # add several at once
print(parties)

parties.discard("FDP")              # remove (no error if it's not there)
print(parties)
````

````python
# Set operations - who took part in BOTH survey waves? Who dropped out?
wave1 = {"Amina", "Boris", "Carla", "Deniz"}
wave2 = {"Boris", "Deniz", "Emma"}

print(wave1 & wave2)    # intersection: in BOTH waves
print(wave1 | wave2)    # union: in AT LEAST one wave
print(wave1 - wave2)    # difference: in wave 1 but NOT in wave 2 (our dropouts!)
````

# Which structure when?

| You want to... | Use a... |
|----------------|----------|
| keep items in order, add/remove freely | **list** |
| protect values from any change | **tuple** |
| look values up by a name/key | **dictionary** |
| count or compare *unique* values | **set** |

Rule of thumb: **lists and dictionaries** will cover 90% of your daily work.

# Importing libraries: giving Python superpowers

Python itself is deliberately small. The magic lives in **libraries** (also called *modules* or *packages*) - collections of ready-made functions written by the community.

Think of libraries as **apps for Python**: need statistics? maps? machine learning? *There's a library for that.*

You activate one with **`import`** - and in Colab, hundreds (including everything we need this week) are **already installed**.

````python
# Import the 'math' library, then use its functions with dot-notation:
import math

print(math.sqrt(144))     # square root
print(math.pi)            # π, more precise than our 3.14159 from this morning
````

````python
# Variant 1: import ONE specific function directly...
from math import sqrt
print(sqrt(144))          # ...then no 'math.' prefix is needed

# Variant 2: import under a short NICKNAME (alias) using 'as'
import math as m
print(m.sqrt(144))
````

````python
# The 'random' library - surprisingly useful in research (sampling!):
import random

countries = ["Germany", "France", "Poland", "Italy", "Spain"]

# randomly choose ONE item (run this cell several times - the answer changes!)
print(random.choice(countries))

# a random whole number, e.g. a dice roll:
print(random.randint(1, 6))

# a random SAMPLE of 2 countries, without replacement:
print(random.sample(countries, 2))
````
