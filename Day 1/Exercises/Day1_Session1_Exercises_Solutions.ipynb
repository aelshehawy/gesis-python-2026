## Exercise 3 - Division, two ways

1. Divide 244 by 5 and print the result.
2. Now print only the **integer part** of that division (no decimals) - remember there is an operator that does exactly this!
3. Print the **remainder** of the division.
4. Add a comment noting which operator does what.

**Solution:**

````python
# Normal division - gives a float
print(244 / 5)     # 48.8

# Integer division: // drops everything after the decimal point
print(244 // 5)    # 48

# Modulo: % gives the REMAINDER of the division
print(244 % 5)     # 4  (because 48 * 5 = 240, and 244 - 240 = 4)
````

## Exercise 4 - Area of a circle

Calculate the area of a circle with radius 3. The formula is: area = π · r²

1. Define the variables you need - `pi` (use `3.14159`) and `r` (set it to 3).
2. Write the equation using the power operator `**`.
3. Make sure your answer is a **float**.
4. Print the result **rounded to 2 decimal places** using an f-string.

**Solution:**

````python
# 1. Define the variables
pi = 3.14159
r = 3

# 2. The formula: area = pi * r squared  (** is the power operator)
area = pi * r ** 2

# 3. pi is a float, so the result is automatically a float - check:
print(type(area))

# 4. Print rounded to 2 decimals using an f-string
print(f"The area of the circle is {area:.2f}")
````

## Exercise 9 - Cleaning messy text

You receive the following messy survey response:

````python
# The code you were given:
answer = "   poLITical sciENCE  "

# 1. Print how many characters the messy string has (spaces count!)
# 2. Clean it: remove the spaces at the edges and make it Title Case - in ONE line, chaining methods
# 3. Save the clean version in a variable called clean_answer and print it
# 4. Print how many characters the clean string has now
````

**Solution:**

````python
answer = "   poLITical sciENCE  "

# 1. Length including the sneaky spaces
print(len(answer))                       # 22

# 2. + 3. Chain the methods: first strip the spaces, then title-case it
clean_answer = answer.strip().title()
print(clean_answer)                      # Political Science

# 4. Length after cleaning
print(len(clean_answer))                 # 17
````

## Exercise 10 - Counting words

Below is a sentence from a famous speech. Using string methods:

1. Count how many times the letter `"e"` appears.
2. Replace `"country"` with `"community"` and print the new sentence.
3. Split the sentence into words and print the resulting list.
4. Print how many **words** the sentence has (use `len()` on your list!).

````python
# The code you were given:
sentence = "Ask not what your country can do for you, ask what you can do for your country"
````

**Solution:**

````python
sentence = "Ask not what your country can do for you, ask what you can do for your country"

# 1. Count occurrences of the letter e
print(sentence.count("e"))

# 2. Replace country → community (replaces ALL occurrences!)
new_sentence = sentence.replace("country", "community")
print(new_sentence)

# 3. Split into a list of words
words = sentence.split()
print(words)

# 4. The number of WORDS is the length of that list
print(f"The sentence has {len(words)} words.")
````

## Exercise 12 (Bonus) - The global health gap

In 2007, life expectancy was **82.6 years in Japan** and **39.6 years in Swaziland** - the highest and lowest values in the dataset we will use on Day 3.

1. Store both values in variables.
2. Calculate the **difference** (in years).
3. Calculate the **ratio** (how many times longer does a person in Japan live?).
4. Print one informative sentence with both numbers, formatting the ratio to 2 decimal places.
5. In a comment: is the ratio an `int` or a `float`? Why?

**Solution:**

````python
# 2007 life expectancies (the extremes of the gapminder data)
life_exp_japan = 82.6
life_exp_swaziland = 39.6

# Difference in years
difference = life_exp_japan - life_exp_swaziland

# Ratio: how many times longer?
ratio = life_exp_japan / life_exp_swaziland

print(f"The gap is {difference:.1f} years - in Japan, people lived {ratio:.2f} times longer than in Swaziland (2007).")

# 5. ratio is a FLOAT: the division operator / in Python ALWAYS returns a float,
#    even when the numbers divide evenly (e.g. 4 / 2 gives 2.0)
print(type(ratio))
````
