## Exercise 2 - Indexing & slicing drills

Using the list below, print:

1. the third city *(careful - which index is that?)*
2. the last two cities, using **one slice**
3. the first three cities, using **one slice**
4. everything **except** the first city, using one slice
5. (Bonus) the string `"Vienna"` has 6 letters - print its first three letters by first grabbing the city from the list, then slicing the string!

```python
# The code you were given:
cities = ["Cologne", "Mannheim", "Vienna", "Zurich", "Amsterdam", "Bergen"]
```

**Solution:**

```python
cities = ["Cologne", "Mannheim", "Vienna", "Zurich", "Amsterdam", "Bergen"]

# 1. The THIRD city sits at index 2 (counting starts at 0!)
print(cities[2])

# 2. Last two - negative start, open end
print(cities[-2:])

# 3. First three - open start, stop BEFORE index 3
print(cities[:3])

# 4. Everything except the first
print(cities[1:])

# 5. (Bonus) First grab the string, then slice it: two operations chained
print(cities[2][:3])    # "Vienna" → "Vie" 
```

## Exercise 3 - The conference guest list

You are organizing a workshop. Start with the list below and perform these operations **in order**, printing the list after each step:

1. `"Farid"` registers → add him to the **end**.
2. `"Aylin"` is a keynote speaker → insert her at the **front** (index 0).
3. `"Boris"` cancels → remove him **by value**.
4. The venue is full - the **last** person in the list unfortunately has to go on the waiting list. Remove them with `.pop()` and save them in a variable called `waitlisted`.
5. Print a message: `Sorry [name], you are on the waiting list!` using an f-string.

```python
# The code you were given:
guests = ["Carla", "Boris", "Deniz", "Emma"]
```

**Solution:**

```python
guests = ["Carla", "Boris", "Deniz", "Emma"]

# 1. Farid registers → append to the end
guests.append("Farid")
print(guests)

# 2. Keynote speaker Aylin → insert at the front
guests.insert(0, "Aylin")
print(guests)

# 3. Boris cancels → remove by value
guests.remove("Boris")
print(guests)

# 4. Venue full → pop() removes AND returns the last person
waitlisted = guests.pop()
print(guests)

# 5. The apology
print(f"Sorry {waitlisted}, you are on the waiting list!")
```

## Exercise 4 - Quick statistics

The list below shows the number of seats a party won in 8 regional parliaments.

1. Print the smallest and largest number of seats.
2. Print the total number of seats across all parliaments.
3. Calculate and print the **average**, formatted to 1 decimal place.
4. Print the list **sorted from largest to smallest**.
5. Check with `in`: did the party win exactly `40` seats anywhere?

```python
# The code you were given:
seats = [23, 41, 8, 33, 27, 41, 15, 30]
```

**Solution:**

```python
seats = [23, 41, 8, 33, 27, 41, 15, 30]

# 1. Extremes
print(min(seats))
print(max(seats))

# 2. Total
print(sum(seats))

# 3. Average = total / count
average = sum(seats) / len(seats)
print(f"Average: {average:.1f} seats")

# 4. Largest → smallest
print(sorted(seats, reverse=True))

# 5. Membership check
print(40 in seats)    # False - 41 twice, but never exactly 40!
```

## Exercise 6 - Your dictionary of capitals

1. Create a dictionary `capitals` where the keys are **four countries** and the values are their **capitals**.
2. Print the capital of one country by looking it up.
3. A fifth country joins - **add** it.
4. Oops, one of your capitals has a typo - **overwrite** it with the corrected name (any change is fine).
5. Print all the **keys**, then all the **values**.
6. Use `in` to check whether `"Brazil"` is in your dictionary.

**Solution:**

```python
# 1. Four countries → capitals
capitals = {"Germany": "Berlin",
            "Egypt": "Cairo",
            "Japan": "Tokyo",
            "Brazil": "Brasilia"}

# 2. Look one up
print(capitals["Egypt"])

# 3. A fifth country joins
capitals["Norway"] = "Oslo"
print(capitals)

# 4. Overwrite an existing value (assigning to an existing key replaces it)
capitals["Brazil"] = "Brasília"     # now with the accent!
print(capitals)

# 5. Keys, then values
print(capitals.keys())
print(capitals.values())

# 6. 'in' checks the keys
print("Brazil" in capitals)
```

## Exercise 7 - Country profiles (nested dictionaries)

Below is a nested dictionary with country profiles.

1. Print Egypt's **whole profile**.
2. Print **only** Japan's capital.
3. Print Germany's **continent**.
4. Add a **new key** `"eu_member"` to Germany's inner dictionary with the value `True`. *(Hint: `profiles["Germany"]["eu_member"] = ...`)*
5. Print Germany's updated profile.
6. (Bonus) Print a sentence via f-string: `Cairo is the capital of Egypt, which has 80.3 million inhabitants.` - pulling **every** piece of information out of the dictionary.

```python
# The code you were given:
profiles = {
    "Germany": {"capital": "Berlin", "population_m": 82.4, "continent": "Europe"},
    "Egypt":   {"capital": "Cairo",  "population_m": 80.3, "continent": "Africa"},
    "Japan":   {"capital": "Tokyo",  "population_m": 127.5, "continent": "Asia"},
}
```

```python
profiles = {
    "Germany": {"capital": "Berlin", "population_m": 82.4, "continent": "Europe"},
    "Egypt":   {"capital": "Cairo",  "population_m": 80.3, "continent": "Africa"},
    "Japan":   {"capital": "Tokyo",  "population_m": 127.5, "continent": "Asia"},
}

# 1. Whole inner dictionary
print(profiles["Egypt"])

# 2. Outer key first, then inner key
print(profiles["Japan"]["capital"])

# 3. Same pattern
print(profiles["Germany"]["continent"])

# 4. Assigning to a new inner key adds it
profiles["Germany"]["eu_member"] = True

# 5. Check the update
print(profiles["Germany"])

# 6. (Bonus) Every value pulled from the dictionary - nothing typed by hand:
print(f"{profiles['Egypt']['capital']} is the capital of Egypt, "
      f"which has {profiles['Egypt']['population_m']} million inhabitants.")
# Note: inside the f-string we use SINGLE quotes for the keys,
# because the f-string itself uses double quotes!
```

## Exercise 8 - Unique values with sets

A (very busy) respondent listed every party they have ever voted for:

1. How many entries does the list have?
2. Create a set of the **unique** parties and print it.
3. How many **different** parties are there?
4. The respondent remembers also voting for `"Pirates"` once - add it to the set.
5. What happens if you `.add("SPD")` again? Print the set and explain in a comment.

```python
# The code you were given:
all_votes = ["SPD", "CDU", "SPD", "Greens", "SPD", "FDP", "Greens", "CDU", "SPD"]
```

**Solution:**

```python
all_votes = ["SPD", "CDU", "SPD", "Greens", "SPD", "FDP", "Greens", "CDU", "SPD"]

# 1. Total entries
print(len(all_votes))          # 9

# 2. set() removes the duplicates
unique_parties = set(all_votes)
print(unique_parties)

# 3. Number of DIFFERENT parties
print(len(unique_parties))     # 4

# 4. Add a new one
unique_parties.add("Pirates")
print(unique_parties)

# 5. Adding "SPD" again changes NOTHING - sets never contain duplicates,
#    and Python doesn't even complain:
unique_parties.add("SPD")
print(unique_parties)
```

## Exercise 9 - Panel dropouts

Two waves of a panel survey. Using **set operations** (`&`, `|`, `-`):

1. Who participated in **both** waves?
2. Who participated in **at least one** wave?
3. Who **dropped out** after wave 1?
4. Who is **new** in wave 2?

```python
# The code you were given:
wave1 = {"Amina", "Boris", "Carla", "Deniz", "Emma"}
wave2 = {"Boris", "Deniz", "Emma", "Farid", "Grete"}
```

**Solution:**

```python
wave1 = {"Amina", "Boris", "Carla", "Deniz", "Emma"}
wave2 = {"Boris", "Deniz", "Emma", "Farid", "Grete"}

# 1. Intersection - in BOTH waves
print(wave1 & wave2)

# 2. Union - in at least one wave
print(wave1 | wave2)

# 3. In wave1 but not wave2 - our dropouts
print(wave1 - wave2)

# 4. In wave2 but not wave1 - the newcomers
print(wave2 - wave1)
```

## Exercise 10 - Library time!

1. Import the `math` library and print the square root of `1764`.
2. Import the `random` library and print one **random country** from your `my_countries` list of Exercise 1. Run it a few times!
3. Simulate a die: print a random whole number between 1 and 6.
4. Draw a random **sample of 2** countries from your list.
5. (Bonus) Use `random.shuffle(my_countries)` and print the list - what did it do? (Careful: `.shuffle` changes the list itself and returns nothing!)

**Solution:**

```python
import math
import random

my_countries = ["Egypt", "Germany", "Japan", "Brazil", "Norway"]

# 1. Square root
print(math.sqrt(1764))          # 42.0 - the answer to everything

# 2. One random country (different on every run!)
print(random.choice(my_countries))

# 3. A die roll: whole number from 1 to 6 (both included)
print(random.randint(1, 6))

# 4. Sample of 2, without replacement
print(random.sample(my_countries, 2))

# 5. (Bonus) shuffle mixes the list IN PLACE (it returns None, so don't print the call itself!)
random.shuffle(my_countries)
print(my_countries)
```

## Exercise 11 (Bonus) - Slicing puzzles

Slices have an optional third number: `list[start:stop:step]`.

Using **one slice each**, produce from the list below:

1. every **second** number, starting from the first - `[10, 30, 50, 70]`
2. the list **reversed** - `[80, 70, 60, 50, 40, 30, 20, 10]` *(a famous trick - step can be negative!)*
3. every second number **of the reversed list** - `[80, 60, 40, 20]`

```python
# The code you were given:
numbers = [10, 20, 30, 40, 50, 60, 70, 80]
```

**Solution:**

```python
numbers = [10, 20, 30, 40, 50, 60, 70, 80]

# 1. Every second item: step 2
print(numbers[::2])

# 2. The famous reversal trick: step -1 walks backwards through the list
print(numbers[::-1])

# 3. Walk backwards, two at a time
print(numbers[::-2])
```

## Exercise 12 (Bonus) - The grand combination

You receive polling data as a list of tuples - each tuple is `(party, percent)`. It should not be modified, which is exactly why tuples were chosen.

1. Print the first tuple.
2. Print **only the party name** of the first tuple (two-step indexing!).
3. Build a **dictionary** from the data *by hand* (type it out): party → percent.
4. Look up the value for `"Greens"` in your dictionary.
5. The percentages are also available as the plain list `percents` below. Use it to print the **highest** value and the **average**, formatted to 1 decimal.
6. In a comment: why is a *dictionary* more convenient here than the *list of tuples*?

```python
# The code you were given:
poll = [("SPD", 20.5), ("CDU", 29.0), ("Greens", 14.5), ("FDP", 5.0), ("AfD", 17.5), ("Linke", 8.5)]
percents = [20.5, 29.0, 14.5, 5.0, 17.5, 8.5]
```

**Solution:**
