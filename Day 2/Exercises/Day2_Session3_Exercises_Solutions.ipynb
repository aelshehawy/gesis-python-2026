## Exercise 1 - May they vote?

Write an `if` / `elif` / `else` chain that checks the variable `age` and prints:

- `"You can vote and run for chancellor!"` if age is 40 or above
- `"You can vote!"` if age is 18-39
- `"Not old enough to vote yet."` otherwise

Test it with ages `50`, `25`, and `12` (just change the variable and re-run).

````python
# The code you were given:
age = 25
````

**Solution:**

````python
age = 25

# Python checks top to bottom and runs the FIRST branch that is True.
# That is why we test the >= 40 case first - if we tested >= 18 first,
# a 50-year-old would get the plain "You can vote!" and never reach the chancellor line!
if age >= 40:
    print("You can vote and run for chancellor!")
elif age >= 18:
    print("You can vote!")
else:
    print("Not old enough to vote yet.")
````

## Exercise 3 - Range drills

1. Print the numbers **-10 to -1** using a `for` loop and `range()`.
2. Print the numbers 0 to 100 **in steps of ten** (0, 10, 20, ... 100).
3. Use a loop and the accumulator pattern to compute the **sum of the numbers 1 to 100** (young Gauss allegedly did this in seconds - your loop is even faster).

**Solution:**

````python
# 1. range(-10, 0) produces -10, -9, ..., -1 (stop is NOT included)
for n in range(-10, 0):
    print(n)
````

````python
# 2. The third argument of range() is the step size.
#    Stop must be 101, otherwise 100 itself would be excluded!
for n in range(0, 101, 10):
    print(n)
````

````python
# 3. Accumulator pattern: start at 0, add each number as it passes
total = 0
for n in range(1, 101):
    total += n          # same as: total = total + n

print(total)            # 5050 - Gauss would approve
````

## Exercise 4 - Mountain heights

*(A classic from my previous courses!)* [Wikipedia's list of mountains by elevation](https://en.wikipedia.org/wiki/List_of_mountains_by_elevation) - pick five, or use the dictionary offered below.

1. Create a dictionary with mountain names as **keys** and elevations in meters as **values**.
2. Print just the mountains' **names**, by looping through the keys.
3. Print just the **elevations**, by looping through the values.
4. Print a series of statements like: `Everest is 8848 meters tall.` - by looping through both at once.
5. (Bonus) Print those same statements sorted **alphabetically** by mountain name. *(Hint: `sorted()` works on `.items()` too!)*

````python
# The code you were given:
# You can use these, or pick your own five:
mountains = {"Everest": 8848, "K2": 8611, "Kangchenjunga": 8586,
             "Lhotse": 8516, "Makalu": 8485}
````

**Solution:**

````python
# 1. The dictionary
mountains = {"Everest": 8848, "K2": 8611, "Kangchenjunga": 8586,
             "Lhotse": 8516, "Makalu": 8485}

# 2. Looping over a dictionary directly yields its KEYS
print("The mountains in our data:")
for name in mountains:
    print(name)

# 3. .values() for the elevations
print("\nTheir elevations:")
for elevation in mountains.values():
    print(elevation)

# 4. .items() unpacks both at once
print("\nThe full picture:")
for name, elevation in mountains.items():
    print(f"{name} is {elevation} meters tall.")
````

````python
# 5. (Bonus) sorted() on .items() sorts the pairs alphabetically by key
for name, elevation in sorted(mountains.items()):
    print(f"{name} is {elevation} meters tall.")
````

## Exercise 5 - Above average?

Using the seats list below and **two loops**:

1. First loop (accumulator): compute the **average** number of seats.
2. Second loop (+ `if` + counter): count **how many** parliaments are **above** the average.
3. Print both results in full sentences.

````python
# The code you were given:
seats = [23, 41, 8, 33, 27, 41, 15, 30]
````

**Solution:**

````python
seats = [23, 41, 8, 33, 27, 41, 15, 30]

# 1. First loop: sum everything up, then divide by the count
total = 0
for s in seats:
    # note that this is the same thing as doing total = total + s. 
    # in practice, the accumulator pattern += is more commonly used
    total += s 
    
average = total / len(seats)

# 2. Second loop: count values above the average
count_above = 0
for s in seats:
    if s > average:
        # note that this is the same thing as doing count_above = count_above + s. 
        count_above += 1  # the counter only grows when the condition holds
        

# 3. Report
print(f"The average is {average:.1f} seats.")
print(f"{count_above} of {len(seats)} parliaments are above the average.")
````

## Exercise 6 - Filtering with loop + if

The list below holds (country, life expectancy) **tuples** for 2007.

1. Using a loop with an `if`, build a list `high_life_exp` containing **only the country names** where life expectancy is above 75.
2. Print the list and its length.
3. (Bonus) Do the same in **one line** with a list comprehension. *(Hint: `for country, value in data` also works inside a comprehension!)*

````python
# The code you were given:
data = [("Japan", 82.6), ("Germany", 79.4), ("Poland", 75.6),
        ("Egypt", 71.3), ("India", 64.7), ("Rwanda", 46.2)]
````

**Solution:**

````python
data = [("Japan", 82.6), ("Germany", 79.4), ("Poland", 75.6),
        ("Egypt", 71.3), ("India", 64.7), ("Rwanda", 46.2)]

# 1. Loop version - two loop variables unpack each tuple
high_life_exp = []
for country, value in data:
    if value > 75:
        high_life_exp.append(country)    # we collect the NAME, not the value

# 2.
print(high_life_exp)
print(len(high_life_exp))

# 3. (Bonus) The same as a one-liner:
high_comp = [country for country, value in data if value > 75]
print(high_comp)
````

## Exercise 8 - While: the doubling population

A country has 50 million inhabitants and grows by **2% per year** (multiply by `1.02`).

1. Using a `while` loop, find out **how many years** it takes for the population to **double** (reach 100 million).
2. Print the answer as a sentence, and the final population with 1 decimal place.
3. (Bonus) Wrap your `while` loop so it also handles this: with growth of only 0.5% a year (`1.005`), does the answer surprise you?

**Solution:**

````python
# 1. + 2.
population = 50.0     # millions
years = 0             # our counter

while population < 100:        # keep going until we reach double
    population *= 1.02         # grow 2% (same as population = population * 1.02)
    years += 1                 # one more year has passed

print(f"With 2% growth, the population doubles in {years} years ({population:.1f} million).")
````

````python
# 3. (Bonus) Same logic, growth of only 0.5% per year:
population = 50.0
years = 0

while population < 100:
    population *= 1.005
    years += 1

print(f"With 0.5% growth it takes {years} years!")
# Small differences in growth rates compound into enormous differences over time -
# the same logic explains why small differences in economic growth matter so much.
````

## Exercise 10 - Loop list comprehension

Rewrite each little loop below as a **one-line list comprehension** (keep the loop version as a comment so you can compare):

1. Squares of 1-10.
2. All words of the sentence, lowercased. *(Split first!)*
3. Only the grades above 80.

````python
# The code you were given:
# 1.
squares = []
for x in range(1, 11):
    squares.append(x ** 2)

# 2.
sentence = "The Hand Sanitizer Was Actually Clear Glue"
words_lower = []
for word in sentence.split():
    words_lower.append(word.lower())

# 3.
grades = [88, 94, 100, 99, 70, 98, 55, 20]
passed = []
for g in grades:
    if g > 80:
        passed.append(g)
````

**Solution:**

````python
# 1. Squares of 1-10
squares = [x ** 2 for x in range(1, 11)]
print(squares)

# 2. Split into words, lowercase each
sentence = "The Hand Sanitizer Was Actually Clear Glue"
words_lower = [word.lower() for word in sentence.split()]
print(words_lower)

# 3. Filter: only grades above 80
grades = [88, 94, 100, 99, 70, 98, 55, 20]
passed = [g for g in grades if g > 80]
print(passed)
````

## Exercise 12 (Bonus) - Comprehension finale

1. Find all numbers from 1 to 1000 that **contain the digit 6**. Print how many there are. *(Hint: `"6" in str(number)`)*
2. From the list below, build a new list with the **floats removed** (keep only the integers). *(Hint: `type(x) == int` - or ask Python for `isinstance(x, int)`.)*
3. Count the number of **spaces** in the sentence below - with a comprehension inside `len()` or `sum()`.

````python
# The code you were given:
original_list = [2, 3.75, 0.04, 59.354, 6, 7.7777, 8, 9]
sentence = "Practice Problems to tackle for List Comprehension mastery"
````

**Solution:**

````python
# 1. Numbers containing the digit 6:
#    convert each number to a string, then check if "6" is in it
with_six = [n for n in range(1, 1001) if "6" in str(n)]
print(with_six[:20])                       # peek at the first 20
print(f"{len(with_six)} numbers contain a 6")   # 271!
````

````python
# 2. Keep only the integers
original_list = [2, 3.75, 0.04, 59.354, 6, 7.7777, 8, 9]

only_ints = [x for x in original_list if isinstance(x, int)]
print(only_ints)

# (type(x) == int works too - isinstance is just the more idiomatic check)
````

````python
# 3. Count the spaces: make a 1 for each space character, then sum them
sentence = "Practice Problems to tackle for List Comprehension mastery"

n_spaces = sum([1 for character in sentence if character == " "])
print(n_spaces)

# Alternative without a comprehension - yesterday's .count() method:
print(sentence.count(" "))
# Many roads to Rome!
````
