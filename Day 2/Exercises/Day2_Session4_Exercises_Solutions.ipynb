## Exercise 2 - Circle, meet function

On Day 1 you calculated the area of ONE circle. Never again:

1. Write a function `circle_area(radius)` that **returns** the area (π · r²). Use `math.pi` for a precise π!
2. Add a one-line **docstring**.
3. Use a **loop** to print the areas for radii 1 through 5, each formatted to 2 decimals.

**Solution:**

```python
import math

def circle_area(radius):
    '''Return the area of a circle with the given radius.'''
    return math.pi * radius ** 2

# The loop applies our recipe five times:
for r in range(1, 6):
    print(f"Radius {r} → area {circle_area(r):.2f}")
```

## Exercise 3 - GDP per capita, industrialized

1. Write a function `gdp_per_capita(gdp_billions, population_millions)` that returns GDP per person in dollars. *(Formula: gdp in billions × 1000 / population in millions.)*
2. The dictionary below maps countries to `(gdp_billions, population_millions)` **tuples** (2007 values). Loop over it and print one line per country: `Germany: 41,737 dollars per person` *(hint: `{value:,.0f}` gives you the thousands separator)*.
3. (Bonus) Which country in the dictionary has the **highest** GDP per capita? Let a loop find it - track the best value AND the best country in two variables.

```python
# The code you were given:
economies = {"Germany": (3439, 82.4), "Poland": (580, 38.5),
             "Norway": (345, 4.6), "Egypt": (443, 80.3), "Japan": (4079, 127.5)}
```

**Solution:**

```python
economies = {"Germany": (3439, 82.4), "Poland": (580, 38.5),
             "Norway": (345, 4.6), "Egypt": (443, 80.3), "Japan": (4079, 127.5)}

# 1. The function
def gdp_per_capita(gdp_billions, population_millions):
    '''Return GDP per person in dollars.'''
    return (gdp_billions * 1000) / population_millions

# 2. Loop over the dictionary; each value is a (gdp, pop) tuple we can index into, using the square brackets: tuple[index_of_the_tuple], such as values[0]
for country, values in economies.items():
    result = gdp_per_capita(values[0], values[1]) # here, we have this dictionary called "economies" -> country is the key, and values is the tuple that is associated with this key
    print(f"{country}: {result:,.0f} dollars per person")
```

```python
# 3. (Bonus) Track the best country while looping:

# we start by initializing our values, which will get updated as the for loop proceeds
best_country = None
best_value = 0                      # everything beats zero at the start

for country, values in economies.items():
    result = gdp_per_capita(values[0], values[1]) # this is the same logic as above -> we loop over key:value pairs and each value is a tuple
    if result > best_value:         # have we found a new leader?
        best_value = result         # if yes, remember the value which therefore becomes the new best_value...
        best_country = country      # ...and remember also who achieved it, which becomes the new best_country

print(f"Highest GDP per capita: {best_country} ({best_value:,.0f} dollars)")
```

## Exercise 4 - The classifier, with a twist

1. Write a function `income_group(gdp_pc)` that returns `"high income"` for values of 12000 and above, `"middle income"` for 1000-11999, and `"low income"` below 1000. *(Rough World Bank-style thresholds.)*
2. Test it on the values `41737`, `2698`, and `820`.
3. (Bonus) Add a **default parameter** `symbol=False`. When `True`, the returned strings get `($$$)` / `($$)` / `($)` attached, e.g. `"high income ($$$)"`. Call it once with and once without.

**Solution:**

```python
# 1. + 2. The basic classifier
def income_group(gdp_pc):
    '''Classify GDP per capita into World Bank-style income groups.'''
    if gdp_pc >= 12000:
        return "high income"
    elif gdp_pc >= 1000:
        return "middle income"
    else:
        return "low income"

print(income_group(41737))
print(income_group(2698))
print(income_group(820))
```

```python
# 3. (Bonus) With a default parameter controlling the symbol:
def income_group(gdp_pc, symbol=False):
    '''Classify GDP per capita; optionally attach a dollar marker.'''
    if gdp_pc >= 12000:
        label = "high income"
        marker = "($$$)"
    elif gdp_pc >= 1000:
        label = "middle income"
        marker = "($$)"
    else:
        label = "low income"
        marker = "($)"

    if symbol:                      # only decorate when asked to
        return label + " " + marker
    else:
        return label

print(income_group(41737))                # default: plain
print(income_group(41737, symbol=True))   # with the marker
```

## Exercise 7 - Search the file

Using `country_facts.txt` and a loop:

1. Print only the facts that mention `"life expectancy"`.
2. Count how many facts mention `"population"` (careful - also catch `"Population"`! Lowercase the line before checking).
3. (Bonus) Build a list of just the **country names** - each line starts with the country followed by `:` ... which string method splits a line at a chosen character? Print your list.

**Solution:**

```python
with open("country_facts.txt", "r") as f:
    lines = f.readlines()

# 1. Facts mentioning life expectancy
for line in lines:
    if "life expectancy" in line.lower():
        print(line.strip())
```

```python
# 2. Count mentions of population, case-insensitively
counter = 0
for line in lines:
    if "population" in line.lower():    # lowercase first → catches Population too
        counter += 1

print(f"{counter} facts mention population")
```

```python
# 3. (Bonus) Each line looks like "Japan: Life expectancy..." -
#    .split(":") cuts at the colon; piece [0] is the country name
country_names = []
for line in lines:
    name = line.split(":")[0]
    country_names.append(name)

print(country_names)

# Or, as yesterday's one-liner:
print([line.split(":")[0] for line in lines])
```

## Exercise 9 - CSV by hand

1. Download `nordic_countries.csv` (cell below).
2. Read it with the **csv module** into a list of rows.
3. Print the header row.
4. Loop over the data rows and print: `Oslo is the capital of Norway.` for each country.
5. Build a list of all countries that are **EU members** (last column is `"yes"`/`"no"`) and print it.

```python
# The code you were given:
import urllib.request
DATA_URL = "https://raw.githubusercontent.com/aelshehawy/gesis-python-2026/main/data/"
urllib.request.urlretrieve(DATA_URL + "nordic_countries.csv", "nordic_countries.csv")
print("Downloaded")
```

**Solution:**

```python
# 1. Fetch the file
import urllib.request
import csv
DATA_URL = "https://raw.githubusercontent.com/aelshehawy/gesis-python-2026/main/data/"
urllib.request.urlretrieve(DATA_URL + "nordic_countries.csv", "nordic_countries.csv")

# 2. Read all rows into a list of lists
with open("nordic_countries.csv", "r") as f:
    reader = csv.reader(f)
    rows = []
    for row in reader:
        rows.append(row)

# 3. The header
print(rows[0])

# 4. Data rows: index 0 = country, 1 = capital
for row in rows[1:]:                      # [1:] skips the header
    print(f"{row[1]} is the capital of {row[0]}.")

# 5. Filter on the last column (index -1 works nicely!)
eu_members = []
for row in rows[1:]:
    if row[-1] == "yes":
        eu_members.append(row[0])

print("EU members:", eu_members)
```

## Exercise 10 (Bonus) - The reusable file tool

Combine EVERYTHING from today: write a function `count_word(filename, word)` that:

- opens the file,
- counts in how many **lines** the word appears (case-insensitive!),
- **returns** that count.

Test it on `country_facts.txt` with the words `"population"`, `"Egypt"`, and `"pizza"`.

**Solution:**

```python
def count_word(filename, word):
    '''Return the number of LINES in the file that contain the word (case-insensitive).'''
    with open(filename, "r") as f:
        lines = f.readlines()

    counter = 0
    for line in lines:
        if word.lower() in line.lower():   # lowercase BOTH sides → case-insensitive
            counter += 1
    return counter

print(count_word("country_facts.txt", "population"))
print(count_word("country_facts.txt", "Egypt"))
print(count_word("country_facts.txt", "pizza"))    # 0 - a data gap in the literature
```

## Exercise 11 (Bonus) - Your own CSV, start to finish

1. With `csv.writer`, create `my_countries.csv` with a header `country,continent,visited` and at least four rows about countries you have visited (or want to!) - `visited` being `yes` or `no`.
2. Read your file back with `csv.reader` and print each row.
3. Using a loop with a **counter**, print how many of the countries you have actually visited.
4. Tomorrow, load this same file with pandas and feel the difference.

**Solution:**

```python
import csv

# 1. Write the file - header first, then one row per country
my_data = [["country", "continent", "visited"],
           ["Egypt", "Africa", "yes"],
           ["Germany", "Europe", "yes"],
           ["Japan", "Asia", "no"],
           ["Brazil", "Americas", "no"]]

with open("my_countries.csv", "w", newline="") as f:
    writer = csv.writer(f)
    for row in my_data:
        writer.writerow(row)

# 2. Read it back
with open("my_countries.csv", "r") as f:
    reader = csv.reader(f)
    rows = []
    for row in reader:
        rows.append(row)
        print(row)

# 3. Count the visited ones
visited_count = 0
for row in rows[1:]:                 # skip the header
    if row[2] == "yes":
        visited_count += 1

print(f"I have visited {visited_count} of {len(rows) - 1} countries.")
```
