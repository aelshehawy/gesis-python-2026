# Why control flow?

So far, our code runs **top to bottom, once**. Today we teach Python to:

1. **Make decisions** - *"IF life expectancy is below 60, flag this country"* → `if` / `elif` / `else`
2. **Repeat work** - *"do this for EACH of my 1,704 rows"* → loops

This is the moment programming starts saving you real time: code that processes 10 items can process 10 million.

# Conditional statements: `if`, `elif`, `else`

The `if` statement runs code **only when a condition is `True`**:

```python
if condition:
    do_something
```

Two things to burn into memory:
1. The **colon `:`** at the end of the `if` line.
2. The **indentation** - the indented lines "belong" to the `if`. Python is strict about this! Colab indents for you automatically after a colon (4 spaces is the convention).

````python
# The condition uses the comparison operators from Day 1:
age = 20

if age >= 18:
    print("You are allowed to vote!")

print("This line is NOT indented - it runs no matter what.")
````

````python
# Change age to 15 and run again - what happens to each print?
age = 15

if age >= 18:
    print("You are allowed to vote!")

print("This line is NOT indented - it runs no matter what.")
````

````python
# 'else' catches every case where the condition was False:
age = 15

if age >= 18:
    print("You are allowed to vote!")
else:
    print("Sorry, not old enough yet.")
````

````python
# 'elif' (= "else if") checks FURTHER conditions, top to bottom.
# Python runs the FIRST branch that is True - and skips all the rest.
life_expectancy = 71.3    # Egypt, 2007

if life_expectancy >= 80:
    print("Very high life expectancy")
elif life_expectancy >= 70:
    print("High life expectancy")
elif life_expectancy >= 60:
    print("Medium life expectancy")
else:
    print("Low life expectancy")

# Try it yourself: change the value to 82, then 65, then 45 - follow the logic!
````

````python
# Conditions can be combined with and / or - just like on Day 1:
country = "Norway"
gdp_per_capita = 74859    # dollars, 2007
population_m = 4.6        # millions

if gdp_per_capita > 30000 and population_m < 10:
    print(f"{country} is rich AND small")

if gdp_per_capita > 30000 or population_m > 100:
    print(f"{country} is rich OR huge (at least one is true)")
````

# For loops

A `for` loop repeats a block of code **for each item in a collection**:

```python
for item in my_list:
    do_something_with(item)
```

Read it as English: *"**for** each item **in** my list: do this."*

Same two rules as `if`: **colon**, then **indentation**.

````python
# Without a loop we would copy-paste print() five times... with a loop:
countries = ["Germany", "France", "Poland", "Italy", "Spain"]

for country in countries:      # take each item of the list, one at a time...
    print(country)             # ...and do this with it
````

````python
# 'country' above is the LOOP VARIABLE - a fresh variable that takes each value in turn.
# You can call it anything, but a descriptive name keeps code readable:
for x in ["Germany", "France", "Poland"]:   # x works too...
    print(x)

# ...but 'for country in countries' reads like a sentence. Be kind to your readers!
````

````python
# The loop body can be as many lines as you like - and use the loop variable freely:
countries = ["Germany", "France", "Poland"]

for country in countries:
    n_letters = len(country)
    print(f"{country} has {n_letters} letters")

print("Done! (not indented → runs once, AFTER the loop finishes)")
````

````python
# Strings are sequences too - we can loop over their characters:
for letter in "GESIS":
    print(letter)
````

````python
# range() generates a sequence of numbers to loop over:
for i in range(5):          # 0, 1, 2, 3, 4  - starts at 0, stops BEFORE 5!
    print(i)
````

````python
# range(start, stop) - from start up to (not including) stop:
for year in range(2024, 2027):
    print(f"The year {year}")

# (There is also range(start, stop, step) - e.g. range(0, 100, 10) counts in tens)
````

## The accumulator pattern: collecting results

The single most useful loop pattern in data work:

1. Create an **empty** variable *before* the loop (a `0` counter, or an empty list `[]`).
2. **Update** it inside the loop.
3. After the loop, it holds your result.

You saw `.append()` yesterday - now it becomes powerful.

````python
# Pattern A: counting & summing with a number
seats = [23, 41, 8, 33, 27, 41, 15, 30]

total = 0                  # 1. start at zero
for s in seats:
    total = total + s      # 2. add each value as we pass it

print(total)               # 3. the grand total
# (shortcut: total += s  does exactly the same as  total = total + s)
````

````python
# Pattern B: building a new list with .append()
countries = ["germany", "france", "poland"]

clean_countries = []                          # 1. empty list
for country in countries:
    clean_countries.append(country.title())   # 2. transform + collect

print(clean_countries)                        # 3. a new, cleaned list!
````

````python
# Pattern C: loop + if = FILTERING (the heart of data analysis!)
life_expectancies = [79.4, 43.5, 80.7, 59.4, 82.6, 48.3, 75.6]

high = []                          # will collect values above 75
for value in life_expectancies:
    if value > 75:                 # only when the condition holds...
        high.append(value)         # ...do we keep the value

print(high)
print(f"{len(high)} of {len(life_expectancies)} countries are above 75 years")
````

# Looping through dictionaries

Dictionaries have three loop flavors - `.items()` is the one you will use most:

````python
life_exp = {"Germany": 79.4, "France": 80.7, "Poland": 75.6, "Japan": 82.6}

# Looping over the dictionary directly gives you the KEYS:
for country in life_exp:
    print(country)
````

````python
# .values() gives just the values:
life_exp = {"Germany": 79.4, "France": 80.7, "Poland": 75.6, "Japan": 82.6}

for value in life_exp.values():
    print(value)
````

````python
# .items() gives BOTH - note the TWO loop variables (key, value):
life_exp = {"Germany": 79.4, "France": 80.7, "Poland": 75.6, "Japan": 82.6}

for country, value in life_exp.items():
    print(f"{country}: life expectancy {value} years")
````

````python
# Loops can nest - one loop inside another.
# For each country (outer), we go through each year (inner):
countries = ["Germany", "Egypt"]
years = [1952, 1977, 2007]

for country in countries:            # outer loop
    print(f"--- {country} ---")
    for year in years:               # inner loop: runs COMPLETELY for each country
        print(f"  looking up {country} in {year}...")
````

# While loops

A `for` loop runs *once per item*. A **`while` loop** runs *as long as a condition stays `True`*:

```python
while condition:
    do_something
```

Use it when you **don't know in advance** how many repetitions you need.

> **Danger zone:** if the condition never becomes `False`, the loop runs **forever**. Something inside the loop must change the condition! (If it happens: the stop button next to the cell, or *Runtime → Interrupt execution*.)

````python
# A simple countdown - 'count' changes inside the loop, so it eventually stops:
count = 5

while count > 0:
    print(count)
    count = count - 1     # ← without this line: infinite loop!

print("Liftoff!")
````

````python
# A real question: Egypt had ~80 million people in 2007, growing ~1.8% a year.
# In which year does the population pass 120 million?
population = 80.0     # millions
year = 2007

while population < 120:
    population = population * 1.018    # grow by 1.8%
    year = year + 1                    # move one year forward

print(f"The population passes 120 million in {year} ({population:.1f} million)")
# We had NO idea how many repetitions this needed - perfect job for while!
````

````python
# break = emergency exit: stop the whole loop immediately
# continue = skip the REST of this round, jump to the next item
countries = ["Germany", "France", "STOP", "Poland"]

for country in countries:
    if country == "STOP":
        break                  # leave the loop entirely - Poland is never reached
    print(country)
````

````python
# continue in action: skip missing values (coded as None) instead of crashing
values = [79.4, None, 80.7, None, 75.6]

for v in values:
    if v is None:       # 'is None' is how you test for None
        continue        # skip this one, go straight to the next value
    print(v * 12)       # only real values arrive at this line
# A very common data-cleaning move - remember it for Day 3!
````

# List comprehensions: loops in one line

Python's most beloved shortcut. This pattern -

```python
new_list = []
for item in old_list:
    new_list.append(expression)
```

- collapses into **one readable line**:

```python
new_list = [expression for item in old_list]
```

Same result, less code. Let's see them side by side.

````python
numbers = [1, 2, 3, 4, 5, 6]

# The loop way (4 lines):
doubled_loop = []
for n in numbers:
    doubled_loop.append(n * 2)
print(doubled_loop)

# The list comprehension way (1 line):
doubled_comp = [n * 2 for n in numbers]
print(doubled_comp)

# → identical results!
````

````python
# Works with any transformation - here: cleaning text
countries = ["  germany ", "FRANCE", " poLand  "]

clean = [c.strip().title() for c in countries]
print(clean)
````

````python
# Add an 'if' at the END to FILTER while you build:
grades = [88, 94, 100, 99, 70, 98, 55, 20]

# keep only the passing grades (above 80):
passed = [g for g in grades if g > 80]
print(passed)

# Read it aloud: "give me g, for each g in grades, IF g is above 80" 
````

````python
# Comprehensions love range() - the first ten square numbers:
squares = [x ** 2 for x in range(1, 11)]
print(squares)
````

**When to use which?**

- Simple *transform* or *filter* → **list comprehension** (concise, fast, very "pythonic").
- Multiple steps, `print()`s, conditions with `elif`... → **normal loop** (clearer).

If a comprehension stops being instantly readable - use a loop. Readability wins.

# Putting it all together: mini-analysis

Everything from today in one small, real analysis - life expectancy in 2007 for a handful of countries. *(On Day 3, pandas will do this for 142 countries at once - but now you understand what happens under the hood!)*

````python
# Life expectancy in 2007 (a small sample of the data we will use tomorrow)
life_exp = {
    "Japan": 82.6, "Switzerland": 81.7, "Germany": 79.4, "Poland": 75.6,
    "Egypt": 71.3, "India": 64.7, "Rwanda": 46.2, "Swaziland": 39.6,
}

# --- Step 1: overall statistics, using the accumulator pattern ---
total = 0
for value in life_exp.values():
    total += value
average = total / len(life_exp)
print(f"Average across our {len(life_exp)} countries: {average:.1f} years")

# --- Step 2: classify each country with if/elif/else ---
for country, value in life_exp.items():
    if value >= 80:
        category = "very high"
    elif value >= 70:
        category = "high"
    elif value >= 60:
        category = "medium"
    else:
        category = "low"
    print(f"{country:<12} {value:>5} years   {category}")

# --- Step 3: collect the countries above the average, with a comprehension ---
above_average = [c for c, v in life_exp.items() if v > average]
print(f"\nAbove average: {above_average}")
````
