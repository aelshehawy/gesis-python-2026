# Why functions?

You already *use* functions all the time: `print()`, `len()`, `sorted()`... Today you learn to **write your own**.

**The motivation:** imagine you have 100 datasets and each needs the same cleaning steps.

- Option A: copy-paste your cleaning code 100 times *(and when you find a bug - fix it 100 times!)*
- Option B: write the recipe **once** as a function, then use it 100 times

A function is a **named, reusable recipe**:

```python
def function_name(ingredients):
    # the recipe steps
    return result
```

- **`def`** starts the definition, then the name, parentheses, and our old friend the **colon**
- the indented **body** is the recipe (it does NOT run yet - it is only being defined!)
- **`return`** hands the result back
- calling `function_name(value)` actually **runs** it

````python
# Our very first function - it takes one ingredient (name) and prints a message:
def thank_you(name):
    print(f"You are doing fantastic work, {name}!")
    print("Thank you for attending this course.")

# Defining it produced NO output - Python just memorized the recipe.
# CALLING it runs the recipe:
thank_you("Amina")
````

````python
# And now the whole point: reuse! Same recipe, different ingredient:
thank_you("Boris")
thank_you("Carla")
````

````python
# What happens if we forget the ingredient? Run it and read the error:
thank_you()
````

`missing 1 required positional argument: 'name'` - the function *requires* a `name`. Precise, readable errors: Python tells you exactly what is missing.

**Vocabulary:** `name` is a **parameter** (the placeholder in the recipe); `"Amina"` is an **argument** (the concrete value you pass in). People mix these up constantly - no one will arrest you.

## ↩ `return`: functions that hand back a result

`print()` inside a function only *displays* something. **`return` gives the result back to your program**, so you can store it in a variable and keep working with it. This distinction confuses everyone at first - let's make it crystal clear:

````python
# This function RETURNS its result:
def add_vat(price):
    result = price * 1.19        # add 19% German VAT
    return result                # hand the result back (this also ENDS the function)

# Because it returns, we can CATCH the result in a variable...
gross = add_vat(100)
print(gross)

# ...and calculate with it!
print(gross * 2)
````

````python
# Compare: this one only PRINTS its result...
def add_vat_print(price):
    print(price * 1.19)

caught = add_vat_print(100)   # it prints, but returns nothing...
print(caught)                 # → None. The result was displayed, then lost!

# Rule of thumb: use return. Print only when displaying IS the purpose.
````

````python
# Functions can take SEVERAL parameters - order matters:
def gdp_per_capita(gdp_billions, population_millions):
    # GDP comes in billions, population in millions → result is in dollars per person
    result = (gdp_billions * 1000) / population_millions
    return result

# Germany 2007: GDP ≈ $3,439 billion, population ≈ 82.4 million
germany = gdp_per_capita(3439, 82.4)
print(f"Germany: {germany:,.0f} dollars per person")

# Poland 2007: GDP ≈ $580 billion, population ≈ 38.5 million
poland = gdp_per_capita(580, 38.5)
print(f"Poland: {poland:,.0f} dollars per person")
````

````python
# You can name the arguments when calling - then the order no longer matters
# (and the call becomes wonderfully readable):
print(gdp_per_capita(population_millions=82.4, gdp_billions=3439))
````

````python
# DEFAULT VALUES: give a parameter a fallback with =
def greet(name, language="English"):
    if language == "German":
        return f"Hallo, {name}!"
    elif language == "Arabic":
        return f"Ahlan, {name}!"
    else:
        return f"Hello, {name}!"

print(greet("Amina"))                  # language not given → uses the default
print(greet("Amina", "German"))        # default overridden
print(greet("Amina", "Arabic"))
````

````python
# Good practice: a DOCSTRING right under the def line documents your function.
# (Triple quotes; this is what help() shows!)
def life_exp_category(value):
    '''Classify a life expectancy value (in years) into a category.'''
    if value >= 80:
        return "very high"
    elif value >= 70:
        return "high"
    elif value >= 60:
        return "medium"
    else:
        return "low"

print(life_exp_category(82.6))
print(life_exp_category(46.2))
help(life_exp_category)          # ← your documentation, on demand
````

## Functions + loops: the dream team

Write the logic **once**, apply it to **many** cases - this is the pattern that scales:

````python
# This morning's classification - now as one clean line per country:
life_exp = {"Japan": 82.6, "Germany": 79.4, "Egypt": 71.3,
            "India": 64.7, "Rwanda": 46.2}

for country, value in life_exp.items():
    category = life_exp_category(value)          # the function does the thinking
    print(f"{country}: {value} years → {category}")
````

````python
# One more tool for your kit: functions from libraries, like sleep()
# (pauses your code - essential later in life for polite web scraping!)
from time import sleep

print("Collecting data", end="")
for i in range(3):
    sleep(1)              # wait 1 second
    print(".", end="")    # end="" keeps everything on one line
print(" done!")
````

### One thing to know: what happens in a function, stays in a function

Variables created **inside** a function exist only while it runs (this is called *local scope*):

```python
def add_vat(price):
    result = price * 1.19    # 'result' lives only inside
    return result

add_vat(100)
print(result)                # NameError - 'result' does not exist out here!
```

That is exactly why `return` matters: it is the official door through which results leave the function.

## Reading a file

The classic three steps: **open → read → close**.

```python
file_object = open("filename", "mode")
```

The most important **modes**: `"r"` = read, `"w"` = write ( overwrites!), `"a"` = append.

Our file `country_facts.txt` *(written for this course, based on the Gapminder data you will meet tomorrow)* contains 15 one-line facts about countries - let's open it.

````python
# Step 1: open the file in read mode
facts_file = open("country_facts.txt", "r")

# What did we get? Let's look:
print(facts_file)   # hm - a file OBJECT, not the text itself!
````

````python
# Step 2: .read() pulls the file's entire content into ONE string
facts_file = open("country_facts.txt", "r")
content = facts_file.read()

# Step 3: close the file (frees it up again - always tidy up!)
facts_file.close()

print(type(content))    # a plain string - everything we know about strings applies!
print(content[:200])    # slice: just the first 200 characters
````

## The better way: `with` (a *context manager*)

Forgetting `.close()` happens to everyone. The `with` statement opens the file, gives it to you, **and closes it automatically** when the indented block ends:

```python
with open("filename", "r") as f:
    content = f.read()
```

This is the standard, professional way - we use it from now on.

````python
# Same result as before, but self-closing:
with open("country_facts.txt", "r") as f:
    content = f.read()

print(f"Our file has {len(content)} characters.")
# (Bonus fact: for files with special characters - ä, ü, é, ç, Arabic, ... -
#  it is good practice to add encoding="utf-8" to open(). Colab usually
#  defaults to it, but being explicit never hurts!)
````

````python
# .readlines() splits the content at every line break → a LIST of lines!
with open("country_facts.txt", "r") as f:
    lines = f.readlines()

print(type(lines))          # a list...
print(len(lines))           # ...with 15 items - one per line
print(lines[0])             # the first line (note the invisible \n line break at its end!)
````

````python
# A list means: everything you know about lists and loops applies!
# Print just the first 5 facts, cleaned up:
with open("country_facts.txt", "r") as f:
    lines = f.readlines()

for line in lines[:5]:            # slice: only the first five
    print(line.strip())           # .strip() removes the trailing line break
````

````python
# Loop + if + string methods = searching a file! Which facts mention population?
with open("country_facts.txt", "r") as f:
    lines = f.readlines()

for line in lines:
    if "population" in line.lower():     # .lower() catches "Population" too
        print("", line.strip())
````

## Writing files

Mode `"w"` creates a file (or **overwrites** an existing one - careful!). `.write()` puts text in; unlike `print()`, it does **not** add line breaks - you write `\n` yourself.

````python
# Let's write our first file:
with open("my_findings.txt", "w") as f:
    f.write("My findings, Day 2\n")               # \n = line break
    f.write("Loops save an enormous amount of time.\n")
    f.write("Functions make code reusable.\n")

print("Written! Check the Files sidebar - you can download it from there (three-dot menu).")
````

````python
# Did it work? Read it back:
with open("my_findings.txt", "r") as f:
    print(f.read())
````

````python
# Mode "a" APPENDS to the end instead of overwriting:
with open("my_findings.txt", "a") as f:
    f.write("PS: mode 'a' adds lines at the end!\n")

with open("my_findings.txt", "r") as f:
    print(f.read())

# One warning: Colab's little hard drive is TEMPORARY - files vanish when
# the session ends. Download anything you want to keep (Files sidebar → three-dot menu → Download).
````

# CSV files

**CSV** (= *Comma-Separated Values*) is THE standard format for tabular data - every statistics package and Excel can read and write it.

A CSV is just a **plain text file**: one line per row, values separated by commas. Let's prove that by reading one with the tools we already have:

````python
# Download a small CSV: 5 Nordic countries (compiled for this course)
import urllib.request
urllib.request.urlretrieve(DATA_URL + "nordic_countries.csv", "nordic_countries.csv")

# It is just text! Look at it with .read() like any other file:
with open("nordic_countries.csv", "r") as f:
    print(f.read())
````

See the structure? **First line = column names** (the *header*), every other line = one country. The built-in `csv` library splits those commas for us properly:

````python
import csv

# csv.reader turns each line into a LIST of values (split at the commas)
with open("nordic_countries.csv", "r") as f:
    reader = csv.reader(f)
    rows = []                    # accumulate the rows - yesterday's pattern!
    for row in reader:
        rows.append(row)

# The first row is the header...
print(rows[0])

# ...the rest is data: lists inside a list, like our mini-dataset yesterday
print(rows[1])
print(rows[1][1])    # row 1, column 1 → Denmark's capital
````

````python
# Loop over the data rows (skipping the header with a slice) and report:
for row in rows[1:]:
    country = row[0]
    population = row[2]
    print(f"{country} has {population} million inhabitants")
````

````python
# Writing a CSV works the same way in reverse - csv.writer + .writerow():
results = [["party", "percent"],       # header first
           ["SPD", 20.5],
           ["CDU", 29.0],
           ["Greens", 14.5]]

with open("poll_results.csv", "w", newline="") as f:   # newline="" avoids blank lines
    writer = csv.writer(f)
    for row in results:
        writer.writerow(row)

# Check our work:
with open("poll_results.csv", "r") as f:
    print(f.read())
````
